```python
from itertools import combinations
from bisect import bisect_left

def subset_sums(nums):
    """Generate all possible subset sums of a list."""
    sums = []
    n = len(nums)
    for i in range(n + 1):
        for comb in combinations(nums, i):
            sums.append(sum(comb))
    return sums


def meet_in_the_middle(arr, target):
    n = len(arr)
    left = arr[:n // 2]
    right = arr[n // 2:]

    # Compute all subset sums for both halves
    left_sums = subset_sums(left)
    right_sums = subset_sums(right)
    right_sums.sort()

    # Check if there exists a combination that equals target
    for s in left_sums:
        need = target - s
        # Binary search in sorted right_sums
        idx = bisect_left(right_sums, need)
        if idx < len(right_sums) and right_sums[idx] == need:
            return True
    return False


# ---------------- MAIN PROGRAM ----------------
T = int(input())
for _ in range(T):
    N, S = map(int, input().split())
    arr = list(map(int, input().split()))

    if meet_in_the_middle(arr, S):
        print("YES")
    else:
        print("NO")
```
